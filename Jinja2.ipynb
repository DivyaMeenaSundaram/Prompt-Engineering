```python
from jinja2 import Template
```

```python
template = Template("""
You are a customer feedback analyst.

Analyze the following {{ product }} review:

{{ review }}

Return:
1. Customer sentiment
2. Main issue
3. Product aspects mentioned
""")
```

```python
prompt = template.render(
    product="Laptop",
    review="The laptop is fast and well designed, but the battery life is shorter than expected."
)

print(prompt)
```

```python
template = Template("""
You are a customer feedback analyst.

{% if customer_type == "premium" %}
Provide a detailed analysis.
{% else %}
Provide a concise analysis.
{% endif %}

Customer Review:
{{ review }}
""")
```

```python
prompt = template.render(
    customer_type="premium",
    review="The product is excellent but expensive."
)

print(prompt)
```
