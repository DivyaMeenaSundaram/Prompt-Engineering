```python
# Import the Gemini library
# This is needed only when we are ready to send
# our prompts to the Gemini model.

import google.generativeai as genai
```

```python
# The same customer question will be given to every prompt version.

customer_question = """
I ordered a pair of headphones last week.
They arrived yesterday, but the left side isn't working.
What can I do?
"""
```

```python
# Prompt Version 1
# This is our initial prompt.

prompt_v1 = """
You are a customer service assistant.

Customer query:
{question}

Respond professionally and helpfully.
"""
```

```python
# Prompt Version 2
# The original prompt is retained.
# A new requirement is added: the response
# should be concise and empathetic.

prompt_v2 = """
You are a customer service assistant.

Customer query:
{question}

Respond professionally, concisely, and empathetically.
"""
```

```python
# Prompt Version 3
# More business requirements have now been added.

prompt_v3 = """
You are a customer service assistant.

Customer query:
{question}

Respond professionally, concisely, and empathetically.

Do not promise a refund or replacement unless
the customer's eligibility has been established.

Structure the response as:
1. Acknowledge the problem
2. Explain the next step
3. Mention any information the customer needs to provide
"""
```

```python
# Fill the {question} variable in each prompt
# with the same customer question.

final_prompt_v1 = prompt_v1.format(
    question=customer_question
)

final_prompt_v2 = prompt_v2.format(
    question=customer_question
)

final_prompt_v3 = prompt_v3.format(
    question=customer_question
)
```

```python
# Display all prompt versions.
# No API call is made in this step.

print("===== PROMPT VERSION 1 =====")
print(final_prompt_v1)

print("\n===== PROMPT VERSION 2 =====")
print(final_prompt_v2)

print("\n===== PROMPT VERSION 3 =====")
print(final_prompt_v3)
```

```python
# Enter the Gemini API key only when you are
# ready to generate the final outputs.

API_KEY = input("Enter your Gemini API key: ")

# Configure Gemini using the supplied API key.
genai.configure(api_key=API_KEY)

# Select the Gemini model.
model = genai.GenerativeModel("gemini-3.6-flash")
```

```python
# Generate a response using Prompt Version 1.

response_v1 = model.generate_content(final_prompt_v1)
```

```python
# Generate a response using Prompt Version 2.

response_v2 = model.generate_content(final_prompt_v2)
```

```python
# Generate a response using Prompt Version 3.

response_v3 = model.generate_content(final_prompt_v3)
```

```python
# Display the response generated by each prompt version.

print("\n===== OUTPUT FROM VERSION 1 =====")
print(response_v1.text)

print("\n===== OUTPUT FROM VERSION 2 =====")
print(response_v2.text)

print("\n===== OUTPUT FROM VERSION 3 =====")
print(response_v3.text)
```
