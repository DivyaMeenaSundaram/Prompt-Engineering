```python
from jinja2 import Template

template = Template("""
You are a Walmart customer support assistant.

Answer the customer's question accurately and concisely.

Customer question:
{{ customer_question }}
""")
```

```python
prompt = template.render(
    customer_question="Can I return this product after 45 days?"
)

print(prompt)
```

```python
questions = [
    "Can I return this product after 45 days?",
    "My order says delivered but I haven't received it.",
    "Can I change my delivery address?",
    "Why was I charged twice?"
]
```

```python
for question in questions:

    prompt = template.render(
        customer_question=question
    )

    print(prompt)
    print("-" * 50)
```

```python
from jinja2 import Template

template = Template("""
You are a Walmart customer support assistant.

Use the provided context to answer the customer's question.

Do not invent information that is not present in the context.

Context:
{{ context }}

Customer question:
{{ customer_question }}

Answer:
""")

prompt = template.render(
    context="""
    Eligible products can generally be returned
    according to the applicable return conditions.
    Some categories may have different requirements.
    """,

    customer_question="Can I return an opened laptop after 60 days?"
)

print(prompt)
```

```python
template = Template("""
You are a customer support assistant.

{% if customer_type == "premium" %}
Provide a highly personalized response.
{% else %}
Provide a standard customer support response.
{% endif %}

Customer question:
{{ question }}
""")
```

```python
prompt = template.render(
    customer_type="premium",
    question="Where is my order?"
)

print(prompt)
```

```python
template = Template("""
You are a customer support assistant.

Relevant policies:

{% for policy in policies %}
- {{ policy }}
{% endfor %}

Customer question:
{{ question }}
""")

prompt = template.render(
    policies=[
        "Returns are subject to applicable eligibility conditions.",
        "Some product categories have different return requirements.",
        "Account-specific issues may require support assistance."
    ],
    question="Can I return this product?"
)

print(prompt)
```

```python
from langchain_core.prompts import PromptTemplate

prompt_template = PromptTemplate(
    input_variables=["customer_question", "context"],

    template="""
You are a Walmart customer support assistant.

Use only the provided context to answer the question.

Do not invent information.

Context:
{context}

Customer question:
{customer_question}

Answer:
"""
)
```

```python
prompt = prompt_template.format(
    customer_question="Can I return an opened laptop after 60 days?",

    context="""
    Eligible products can generally be returned
    according to the applicable return conditions.
    Some categories may have different requirements.
    """
)

print(prompt)
```
